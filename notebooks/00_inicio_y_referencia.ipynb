## 1. Qué significa comparar las mismas visitas

Una **visita-persona** representa una visita de una persona al Parque. Su ida, los desplazamientos durante la visita y su regreso generan actividad de transporte, pero no añaden visitas. Si esa persona utiliza varias lanzaderas, sus embarques no se suman como nuevos visitantes. Tampoco se trata de personas únicas durante todo el año: una persona puede visitar el Parque en fechas distintas.

La tabla 1 resume las condiciones de los escenarios. Mantener la misma demanda permite comparar cómo cambia su transporte; no significa que ya esté demostrada la viabilidad de cualquier pico de demanda.

**Tabla 1. Comparación de movilidad representada para 2028.**

| Aspecto | Sin intervención | Con intervención |
|---|---|---|
| Visitas al Parque | 2.500.000 visitas-persona/año | Las mismas 2.500.000 visitas-persona/año |
| Acceso en coche | Reparto de referencia preparado a partir de las fuentes | Cupo ordinario global de 650 coches/día; opciones de punta de 700 o 750 cuando corresponda |
| Servicio de lanzadera del proyecto | No se incorpora el nuevo sistema | Hasta 27 propias programadas en ordinaria, de una dotación de treinta; en punta, 28 o treinta según las condiciones de sustitución |
| Sustitución de lanzaderas | No aplica al nuevo sistema | Al menos tres propias sin programar en ordinaria; dos con 28 programadas; con treinta programadas, dos externas disponibles para sustitución |
| Excursiones | Participación estimada de referencia | La misma participación en la comparación principal; los aumentos se estudian como sensibilidad |
| Servicios de acceso de TITSA 342 y 348 | Se incluyen sus servicios y viajeros de referencia | Se sustituyen por el sistema de lanzaderas; sus viajeros se incorporan al reparto sin añadir visitas |
| Viaje hasta las cabeceras | Según el modo de referencia | Quienes acceden a las lanzaderas llegan en coche a la cabecera; el balance incluye ese acceso y su regreso, también para los antiguos viajeros de TITSA |
| Visita interior de los coches autorizados | En coche | También en coche en este cálculo. Su posible traslado futuro a lanzaderas interiores requiere dimensionamiento y no aporta ahorro adicional al balance actual |

Fuente: condiciones del [libro de hipótesis](../config/parametros_modelo.xlsx), su [índice de referencia](../config/referencia_ordenacion_2028.json) y el [informe entregado](../docs/informe_energetico/INFORME_ENERGETICO_v0.28_2026-11-16.pdf). TITSA es Transportes Interurbanos de Tenerife. Una **cabecera** es el lugar de intercambio entre el coche de acceso y la lanzadera; una **lanzadera** es el autobús del sistema propuesto, de acceso o interior.

**27 es un máximo ordinario, no una flota fija que circule todos los días.** Las unidades programadas prestan servicio; las de sustitución permiten reemplazarlas y no se cuentan automáticamente como autobuses comerciales adicionales. La programación por fecha y la atención completa de ida y regreso se explicarán en 04. Esta tabla presenta condiciones del estudio; no acredita autorizaciones ni disponibilidad contractual.

## 5. Leer las condiciones y su procedencia

El **único editor de hipótesis** es [config/parametros_modelo.xlsx](../config/parametros_modelo.xlsx). La tabla 4 se obtiene de la instantánea comprobada con ese libro: muestra las 31 condiciones efectivas, su unidad, el motivo registrado y la celda exacta de procedencia. Las unidades de energía que figuran en la configuración son megavatios hora (MWh) y toneladas equivalentes de petróleo (tep); su conversión se conserva en el libro, aunque aquí no se calcule energía. Una fracción de 0,10 equivale al 10 % de la magnitud indicada en su fila.

Los otros archivos de configuración contienen reglas técnicas, correspondencias o huellas de la referencia. La **huella SHA-256** es una cadena calculada a partir de los bytes de un archivo: permite detectar cambios. No es una validación institucional del dato. El índice sellado comprueba que se está reproduciendo esta referencia; cambiar hipótesis requiere documentar y comprobar otra referencia, no editar huellas para aceptar una discrepancia.

El lector también conserva una capa base de compatibilidad con preparaciones anteriores. No utilizo sus valores como otro escenario activo: las condiciones efectivas son las que se presentan aquí. En 01 explico además el libro original de preparación de aforos, que es una fuente inmutable en Bronze, no otro editor.

## 6. Qué comprueba la apertura

La apertura coteja las condiciones del Excel, las entradas y el código de la ejecución elegida. Si ya hay etapas completadas, comprueba también los archivos que aceptaron. La tabla 5 resume el catálogo por función; no presupone que cada archivo se haya usado numéricamente en todas las etapas.

Las **entradas de cálculo** permiten construir resultados. Las **referencias para comparar** permiten contrastarlos después. Las referencias Silver y Gold se catalogan por separado; su verificación completa se solicita expresamente cuando procede. Comprobar la huella de una referencia no equivale a utilizar sus cifras para fabricar un resultado.


```python
fuentes = pd.DataFrame(DATOS["registro"]["fuentes"])
roles = {"codigo": "Código", "configuracion": "Configuración",
         "definicion_del_entorno": "Definición del entorno",
         "contrato_de_preparacion": "Contrato de preparación",
         "entrada_de_calculo": "Entrada de cálculo",
         "referencia_para_comparar": "Referencia para comparar después"}
resumen_fuentes = fuentes.groupby("rol", as_index=False).agg(Archivos=("ruta", "size"))
resumen_fuentes["Función"] = resumen_fuentes["rol"].map(roles)
display(resumen_fuentes[["Función", "Archivos"]].style.set_uuid("inicio_fuentes").hide(axis="index")
        .set_caption("Tabla 5. Archivos catalogados por función. Fuente: registro inicial de la ejecución; no representa una descarga nueva."))
print("Archivos comprobados al abrir:", DATOS["comprobacion_actual"]["archivos_comprobados"])
print("Referencias catalogadas cuya huella no se ha releído en esta apertura:",
      DATOS["comprobacion_actual"]["referencias_catalogadas_sin_leer"])
print("Entorno coincidente con el de creación:", "Sí" if DATOS["mismo_entorno"] else "No; revisar versiones antes de calcular")
```

## 7. Diferenciar la preparación de los resultados

**Preparar un registro no acredita un cálculo.** El registro inicial conserva su estado de preparación; el historial de etapas indica lo que ocurrió después. La tabla 6 consulta ese historial desde disco. Una etapa solo aporta productos utilizables cuando está completada; un fallo o una interrupción conserva su intento, pero no lo convierte en resultado válido.

Del mismo modo, solicitar reproducción no demuestra que se haya ejecutado COPERT. En 05, `consultar` utiliza exportaciones distribuidas y comprueba sus factores; `ejecutar_motor` inicia el motor instalado. Solo completar esta última acción acredita una ejecución nueva del motor. La ejecución distribuida de consulta permite reconstruir el balance usando sus factores identificados, sin atribuirle otro origen.


```python
filas = []
for etapa, estado in DATOS["estado_etapas"]["etapas"].items():
    filas.append({"Paso": etapa, "Materia": estado["nombre"], "Estado efectivo": estado["estado"],
                  "Acción registrada": estado.get("accion", "No iniciada"),
                  "Archivos aceptados": len(estado.get("productos", []))})
display(pd.DataFrame(filas).style.set_uuid("inicio_estados").hide(axis="index")
        .set_caption("Tabla 6. Estado efectivo de las etapas de la ejecución seleccionada. Fuente: historial verificado al abrir; no se ejecutan estas etapas al mostrar la tabla."))
print("Balance calculado en esta ejecución:", "Sí" if RESUMEN["balance_calculado_en_esta_ejecucion"] else "No")
print("Motor COPERT ejecutado en esta ejecución:", "Sí" if RESUMEN["motor_copert_ejecutado_en_esta_ejecucion"] else "No")
```
